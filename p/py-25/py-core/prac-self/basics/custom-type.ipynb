```python
# The vars() function returns the __dict__ attribute of an object.

# The __dict__ attribute is a dictionary containing the object's
# changeable attributes.

# Note: calling the vars() function without parameters will return
# a dictionary containing the local symbol table.
```

```python
class Coordinate:
  x = 10
  y = -5
  z = 0

  def __str__(self):
    return '{} {} {}'.format(self.x, self.y, self.z)


print(vars(Coordinate))

point1 = Coordinate()
print(point1)
print(type(point1))

a = vars(point1)
print(type(a))
print(a)
```
